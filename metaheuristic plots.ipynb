```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ------------------ CONFIG ------------------
CSV_PATH = r"C:\Users\igork\Downloads\uav_export_2\path_evolve.csv"
OUT_DIR = Path("plots")
OUT_DIR.mkdir(exist_ok=True)

DPI = 200
LINE_WIDTH = 2
FONT_SIZE = 12
# --------------------------------------------

# Load CSV
df = pd.read_csv(CSV_PATH)
df = df.sort_values("generation")

plt.rcParams.update({
    "font.size": FONT_SIZE,
})

# ---------- Cost per generation ----------
plt.figure(figsize=(7, 4))
plt.plot(
    df["generation"],
    df["cost"],
    linewidth=LINE_WIDTH,
)
plt.xlabel("Generation")
plt.ylabel("Objective value")
plt.title("Objective value per Generation")
plt.grid(True)
plt.tight_layout()

cost_path = OUT_DIR / f"{df["algo"]}cost_per_generation.pdf"
plt.savefig(cost_path, dpi=DPI)
plt.close()

# ---------- Distance per generation ----------
plt.figure(figsize=(7, 4))
plt.plot(
    df["generation"],
    df["total_distance_m"],
    linewidth=LINE_WIDTH,
)
plt.xlabel("Generation")
plt.ylabel("Route length [m]")
plt.title("Distance per Generation")
plt.grid(True)
plt.tight_layout()

plt.savefig(OUT_DIR / f"{df["algo"]}distance_per_generation.pdf")
plt.close()
print(f"Saved plots to: {OUT_DIR.resolve()}")
```
